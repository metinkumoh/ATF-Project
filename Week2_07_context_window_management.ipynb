```python
import json

with open("/content/chunks.json", "r", encoding="utf-8") as f:
    chunks = json.load(f)

print("✅ Chunks loaded successfully!")
print("Number of chunks:", len(chunks))
print("First chunk:", str(chunks[0])[:500])
```

```python

from context_manager import ContextManager

# Create the context manager
manager = ContextManager(
    max_tokens=1000,
    system_reserve=100,
    response_reserve=200,
    history_limit=10,
    debug=True
)

# Import your Advanced 1 chunks
total_added = manager.ingest_chunks(chunks)

print("✅ Advanced 1 successfully connected to Week 2.2!")
print("Chunks imported:", total_added)
print("Total knowledge entries:", len(manager.knowledge))
```

```python

# Test knowledge retrieval
question = "What running activities are recorded?"

result = manager.process(question, top_k=3)

print("QUESTION:", question)

print("\nKnowledge chunks included:",
      result["knowledge_included"])

print("\nHistory messages included:",
      result["history_included"])

print("\nLLM-READY CONTEXT")
print("=" * 60)
print(result["prompt"])

print("\nTOKEN USAGE")
print("Before:", result["before_tokens"])
print("After:", result["after_tokens"])
print("Available input budget:", result["input_budget"])
```

```python

import matplotlib.pyplot as plt

questions = [
    "What running activities are recorded?",
    "What duration is recorded for running?",
    "What information does the knowledge base contain?"
]

for question in questions:
    report = manager.process(question, top_k=3)

    print(
        question,
        "| Before:", report["before_tokens"],
        "| After:", report["after_tokens"]
    )

# Visualize real context processing results
manager.visualize()
plt.show()

print("✅ Real token usage visualization generated!")
```

```python
print("FINAL INTEGRATION CHECK")
print("=" * 50)

assert len(chunks) > 0, "No Advanced 1 chunks loaded"
assert len(manager.knowledge) > 0, "Knowledge base is empty"
assert len(manager.reports) > 0, "No processing reports"

for report in manager.reports:
    assert report["after_tokens"] <= report["input_budget"], \
        "Context exceeded the estimated input budget"

print("✅ Advanced 1 document chunks loaded")
print("✅ Knowledge base connected")
print("✅ Context processing completed")
print("✅ Estimated token budgets respected")
print("✅ Actual processing metrics generated")

print("\n🏆 Final integration checks passed!")
```
